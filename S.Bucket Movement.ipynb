```python
import numpy as np
import pandas as pd
```

```python
# Import Data
start_date='2025-08-01'
end_date='2025-08-31'
month='Aug25'
Prayas_cols=[' FINPAGE_LOAN_NO','OD_DAYS','PRINCIPAL_TOTAL',' OUTSTANDING_PRINCIPAL',' STATUS',' LOAN_CLOSURE_DATE']
#Ananya_cols=[' LOAN_NO','OD_DAYS',' PRINCIPAL_TOTAL',' OUTSTANDING_PRINCIPAL','STATUS',' LOAN_CLOSURE_DATE']
Pre=pd.read_csv(r"D:\Sonali\Portfolio Jan-Aug25\Portfolio Report Jul25.csv",usecols=Prayas_cols)
cur=pd.read_csv(r"D:\Sonali\Portfolio Jan-Aug25\Portfolio Report Aug25.csv",usecols=Prayas_cols)
```

```python
cur.shape
```

```python
# Replace values "'" to ""
cur[' FINPAGE_LOAN_NO']=cur[' FINPAGE_LOAN_NO'].apply(lambda x: x.replace("'",""))
Pre[' FINPAGE_LOAN_NO']=Pre[' FINPAGE_LOAN_NO'].apply(lambda x: x.replace("'",""))
```

```python
# change data type 
Pre[' FINPAGE_LOAN_NO']=Pre[' FINPAGE_LOAN_NO'].astype(object)
cur[' FINPAGE_LOAN_NO']=cur[' FINPAGE_LOAN_NO'].astype(object)
```

```python
# merge data
Cur_pre=pd.merge(cur,Pre[[' FINPAGE_LOAN_NO','OD_DAYS']],on=' FINPAGE_LOAN_NO',how='left',suffixes=('_Cur','_Pre'))
```

```python
# delete lag or leading unwanted space from columns name
Cur_pre.columns=Cur_pre.columns.str.strip()
```

```python
Cur_pre['LOAN_CLOSURE_DATE'].isna().sum()
```

```python
Cur_pre['LOAN_CLOSURE_DATE']=pd.to_datetime(Cur_pre['LOAN_CLOSURE_DATE'],errors='coerce')
# Convert LOAN_CLOSURE_DATE to datetime if not already
Cur_pre['LOAN_CLOSURE_DATE'] = pd.to_datetime(Cur_pre['LOAN_CLOSURE_DATE'], dayfirst=True, errors='coerce')

# Filter rows
filter_data = Cur_pre[(Cur_pre['STATUS'].isin(['Active', 'Death']))]
```

```python
Cur_pre['LOAN_CLOSURE_DATE'].isna().sum()
```

```python
# create new columns base on multiple condition


conditions = [
     filter_data['OD_DAYS_Pre'] == 0,
    (filter_data['OD_DAYS_Pre'] > 0) & (filter_data['OD_DAYS_Pre'] <= 30),
    (filter_data['OD_DAYS_Pre'] > 30) & (filter_data['OD_DAYS_Pre'] <= 60),
    (filter_data['OD_DAYS_Pre'] > 60) & (filter_data['OD_DAYS_Pre'] <= 90),
    (filter_data['OD_DAYS_Pre'] > 90) & (filter_data['OD_DAYS_Pre'] <= 120),
    (filter_data['OD_DAYS_Pre'] > 120) & (filter_data['OD_DAYS_Pre'] <= 150),
    (filter_data['OD_DAYS_Pre'] > 150) & (filter_data['OD_DAYS_Pre'] <= 180),
     filter_data['OD_DAYS_Pre'] > 180
]

choices = ['B0', 'B1', 'B2', 'B3', 'B4', 'B5', 'B6', 'B7']

filter_data['DPD_Range_Pre'] = np.select(conditions, choices, default='B0')

conditions1 = [
     filter_data['OD_DAYS_Cur'] == 0,
    (filter_data['OD_DAYS_Cur'] > 0) &  (filter_data[  'OD_DAYS_Cur'] <= 30),
    (filter_data['OD_DAYS_Cur'] > 30) & (filter_data[ 'OD_DAYS_Cur'] <= 60),
    (filter_data['OD_DAYS_Cur'] > 60) & (filter_data[ 'OD_DAYS_Cur'] <= 90),
    (filter_data['OD_DAYS_Cur'] > 90) & (filter_data[ 'OD_DAYS_Cur'] <= 120),
    (filter_data['OD_DAYS_Cur'] > 120) & (filter_data['OD_DAYS_Cur'] <= 150),
    (filter_data['OD_DAYS_Cur'] > 150) & (filter_data['OD_DAYS_Cur'] <= 180),
     filter_data['OD_DAYS_Cur'] > 180
]

choices1 = ['B0', 'B1', 'B2', 'B3', 'B4', 'B5', 'B6', 'B7']

filter_data['DPD_Range_Cur'] = np.select(conditions1, choices1, default=None)
```

```python
filter_data['OUTSTANDING_PRINCIPAL'] =filter_data['OUTSTANDING_PRINCIPAL'] / 100000
```

```python
filter_data.shape
```

### Laon Count

```python
filter_data['DPD_Range_Cur'].unique()
```

```python
# create Pivot or summary table with custome columns
Bucket_Count=pd.pivot_table(data=filter_data,
               index='DPD_Range_Pre',
               columns='DPD_Range_Cur',
               values='FINPAGE_LOAN_NO',
               aggfunc='count').reset_index()

# Create a new row with 'Loan_Count' in all columns
count_row = pd.DataFrame([['Loan_Count'] * Bucket_Count.shape[1]], columns=Bucket_Count.columns)

# Concatenate the new row at the top
Bucket_Count = pd.concat([count_row, Bucket_Count], ignore_index=True)
# replace Value
Bucket_Count.at[0, 'DPD_Range_Pre']=Bucket_Count.at[0, 'DPD_Range_Pre'].replace('Loan_Count','Bucket')
Bucket_Count
```

### POS SUM

```python
Bucket_Sum=pd.pivot_table(data=filter_data,
               index='DPD_Range_Pre',
               columns='DPD_Range_Cur',
               values='OUTSTANDING_PRINCIPAL',
               aggfunc='sum').reset_index()

# Create a new row with 'POS' in all columns
pos_row = pd.DataFrame([['POS'] * Bucket_Sum.shape[1]], columns=Bucket_Sum.columns)

# Concatenate the new row at the top
Bucket_Sum = pd.concat([pos_row, Bucket_Sum], ignore_index=True)
# replace Value
Bucket_Sum.at[0, 'DPD_Range_Pre']=Bucket_Sum.at[0, 'DPD_Range_Pre'].replace('POS', 'Bucket')
Bucket_Sum.head()
```

```python
Condition=[(filter_data['DPD_Range_Cur'].str.strip().str[-1] == filter_data['DPD_Range_Pre'].str.strip().str[-1]),
           (filter_data['DPD_Range_Cur'].str.strip().str[-1] > filter_data['DPD_Range_Pre'].str.strip().str[-1]),
           (filter_data['DPD_Range_Cur'].str.strip().str[-1] < filter_data['DPD_Range_Pre'].str.strip().str[-1])]

Choice=["SB","RF","RB",]

filter_data['DPD Status']=np.select(Condition,Choice,default="")
```

```python
filter_data['DPD Status'].unique()
```

```python
# DPD Move for Count Loan (Roll forward / Bckward/Stable)
Loan_DPD_Move=pd.pivot_table(data=filter_data,
               index='DPD_Range_Pre',
               columns='DPD Status',
               values='FINPAGE_LOAN_NO',
               aggfunc='count',
               margins=True).reset_index()

Count_Denominator=filter_data['FINPAGE_LOAN_NO'].count()
Loan_DPD_Move.insert(2,'RB%',((Loan_DPD_Move['RB']/Count_Denominator)))
Loan_DPD_Move.insert(4,'RF%',((Loan_DPD_Move['RF']/Count_Denominator)))
Loan_DPD_Move.insert(6,'SB%',((Loan_DPD_Move['SB']/Count_Denominator)))
Loan_DPD_Move
```

```python
# DPD Move for POS
POS_DPD_Move=pd.pivot_table(data=filter_data,
               index='DPD_Range_Pre',
               columns='DPD Status',
               values='OUTSTANDING_PRINCIPAL',
               aggfunc='sum',
               margins=True).reset_index()

POS_Denominator=filter_data['OUTSTANDING_PRINCIPAL'].sum()
POS_DPD_Move.insert(2,'RB%',((POS_DPD_Move['RB']/POS_Denominator)))
POS_DPD_Move.insert(4,'RF%',((POS_DPD_Move['RF']/POS_Denominator)))
POS_DPD_Move.insert(6,'SB%',((POS_DPD_Move['SB']/POS_Denominator)))
POS_DPD_Move
```

```python
# Export data according how you want in excel sheet
with pd.ExcelWriter(r'D:\Sonali\Bucket_Movement Ananya'+month+'.xlsx', engine='xlsxwriter') as writer:
    workbook = writer.book
    worksheet = workbook.add_worksheet('All_Reports')
    writer.sheets['All_Reports'] = worksheet

    # ======= LEFT SIDE BLOCK (VERTICAL STACK) =======
    start_row = 0
    start_col = 0

    # Write Report 1: Bucket_Count
    Bucket_Count.to_excel(writer, sheet_name='All_Reports', startrow=start_row, startcol=start_col, index=False)
    start_row += len(Bucket_Count) + 3  # Add spacing

    # Write Report 2: Bucket_Sum (below Bucket_Count)
    Bucket_Sum.to_excel(writer, sheet_name='All_Reports', startrow=start_row, startcol=start_col, index=False)

    # ======= RIGHT SIDE BLOCK (VERTICAL STACK) =======
    start_row = 0  # reset row for right side
    start_col = Bucket_Count.shape[1] + 1  # Add spacing columns (e.g., 5 blank columns)

    # Write Report 3: Loan_DPD_Move (top-right)
    Loan_DPD_Move.to_excel(writer, sheet_name='All_Reports', startrow=start_row, startcol=start_col, index=False)
    start_row += len(Loan_DPD_Move) + 3

    # Write Report 4: POS_DPD_Move (below Loan_DPD_Move)
    POS_DPD_Move.to_excel(writer, sheet_name='All_Reports', startrow=start_row, startcol=start_col, index=False)
```
